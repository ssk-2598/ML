```python
import pandas as pd
from sklearn.ensemble import RandomForestClassifier
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score, classification_report

RANDOM_STATE = 42

# ---------------------------------------------------------------------
# 1. Load the simple dataset
# ---------------------------------------------------------------------
df = pd.read_csv("placement_predict_50k_adjusted.csv")

print("Sample of the dataset:")
print(df.head())
print(f"\nDataset shape: {df.shape}")
print(f"Class balance:\n{df['PlacementStatus'].value_counts()}")

TARGET = "PlacementStatus"
DROP_COLS = ["StudentID"]  # just an id, not a predictor

feature_names = [c for c in df.columns if c not in [TARGET] + DROP_COLS]

# Identify categorical columns for one-hot encoding
categorical_cols = df[feature_names].select_dtypes(include=['object']).columns
df = pd.get_dummies(df, columns=categorical_cols, drop_first=True) # Apply one-hot encoding

# Update feature names after one-hot encoding
feature_names = [c for c in df.columns if c not in [TARGET] + DROP_COLS]

X = df[feature_names].values
y = df[TARGET].values

# ---------------------------------------------------------------------
# 2. Train / test split
# ---------------------------------------------------------------------
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.2, random_state=RANDOM_STATE
)

# ---------------------------------------------------------------------
# 3. Train a Random Forest
# ---------------------------------------------------------------------
rf = RandomForestClassifier(
    n_estimators=100,
    max_features="sqrt",
    oob_score=True,
    random_state=RANDOM_STATE,
)
rf.fit(X_train, y_train)

# ---------------------------------------------------------------------
# 4. Evaluate
# ---------------------------------------------------------------------
y_pred = rf.predict(X_test)
test_accuracy = accuracy_score(y_test, y_pred)

print(f"\nOOB score: {rf.oob_score_:.4f}")
print(f"Test accuracy: {test_accuracy:.4f}")
print("\nClassification report:")
print(classification_report(y_test, y_pred))

# ---------------------------------------------------------------------
# 5. Feature importances
# ---------------------------------------------------------------------
print("Feature importances:")
for name, importance in sorted(
    zip(feature_names, rf.feature_importances_), key=lambda x: -x[1]
):
    print(f"  {name}: {importance:.4f}")
```
