```python
import pandas as pd
df=pd.read_csv("placement_predict_50k.csv")
import matplotlib.pyplot as plt
import seaborn as sns
plt.figure(figsize=(6,4))

sns.countplot(data=df, x="Stream")

plt.title("Placement Count by Stream")
plt.xlabel("Stream")
plt.ylabel("Count")

plt.show()

plt.figure(figsize=(6,4))

plt.hist(df["CGPA"].dropna(),
         bins=15,
         color="skyblue",
         edgecolor="black")

plt.title("CGPA Distribution")
plt.xlabel("CGPA")
plt.ylabel("Frequency")

plt.show()

# Simplify the heatmap by showing correlation of features with 'PlacementStatus' as a bar plot
plt.figure(figsize=(10, 6))

numeric_df = df.select_dtypes(include="number")

# Calculate correlations with 'PlacementStatus'
correlations = numeric_df.corr()['PlacementStatus'].drop('PlacementStatus').sort_values(ascending=False)

sns.barplot(x=correlations.values, y=correlations.index, palette='viridis')
plt.title("Correlation with PlacementStatus")
plt.xlabel("Correlation Coefficient")
plt.ylabel("Feature")
plt.show()
```
